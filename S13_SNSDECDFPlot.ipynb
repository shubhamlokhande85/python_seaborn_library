4.ECDfplot


An ECDF plot (Empirical Cumulative Distribution Function plot) is a graph that shows how data values accumulate. It is a simple and powerful way to understand the distribution of a dataset without grouping values into bins like a histogram


ECDF (Empirical Cumulative Distribution Function):
An ECDF plot is a graph that shows the proportion or percentage of observations that are less than or equal to each value in a dataset.



Empirical - means it is based on the actual observed data.



Cumulative -  means the values are added up progressively.


Distribution - Function describes how the data are spread.


Syntax 

sns.ecdfplot(
    data=None,
    x=None,
    y=None,
    hue=None,
    weights=None,
    stat="proportion",
    complementary=False,
    palette=None,
    hue_order=None,
    log_scale=None,
    legend=True,
    ax=None
)

where,



| Parameter            | Meaning                                                                  | Example                        |
| -------------------- | ------------------------------------------------------------------------ | ------------------------------ |
| `data`               | Dataset used for plotting                                                | `data=df`                      |
| `x`                  | Variable for the x-axis                                                  | `x="age"`                      |
| `y`                  | Variable for the y-axis (optional)                                       | `y="salary"`                   |
| `hue`                | Separates data into groups using different colors                        | `hue="gender"`                 |
| `weights`            | Gives different importance to observations                               | `weights="population"`         |
| `stat`               | Shows results as `"proportion"` (0–1), `"percent"` (0–100), or `"count"` | `stat="percent"`               |
| `complementary=True` | Displays the complementary ECDF (proportion greater than a value)        | `complementary=True`           |
| `palette`            | Color scheme                                                             | `palette="Set2"`               |
| `hue_order`          | Controls the order of grouped categories                                 | `hue_order=["Male", "Female"]` |
| `log_scale`          | Uses a logarithmic axis                                                  | `log_scale=True`               |
| `legend`             | Shows or hides the legend                                                | `legend=False`                 |
| `ax`                 | Draws the plot on a specified Matplotlib axis                            | `ax=ax`                        |

1.import all required libraries

```python
import numpy as np
import pandas as pd 
import matplotlib.pyplot as plt
import seaborn as sns 
```

2. import dataset and specific observation 

```python
df=pd.read_excel("C:/data_analysis_using_python/supermarket_sales.xlsx")
df=df[["Gender","Unit price","gross income","Quantity","Payment","Total"]]
df.head()
```

3.Create basic ECDFplot 

```python
sns.ecdfplot(data=df,x="Unit price")
```

4.Group the ECDFplot

```python
sns.ecdfplot(data=df, x ="Unit price", hue="Gender",stat="count")
```

5.Change the axis 

```python
sns.ecdfplot(data=df,y="Unit price", hue="Gender")
```
