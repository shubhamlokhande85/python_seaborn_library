B.Distribution plots


2.KDEPLOT



kdeplot() (Kernel Density Estimate Plot) is a Seaborn function used to visualize the probability distribution of continuous data.



It creates a smooth curve that represents the distribution of values. It is similar to a histogram but provides a smoother estimate of the data distribution


Syntax


sns.kdeplot(
    data=None,
    x=None,
    y=None,
    hue=None,
    fill=False,
    multiple='layer',
    bw_adjust=1,
    bw_method=None,
    cut=2,
    clip=None,
    cumulative=False,
    common_norm=True,
    common_grid=False,
    levels=10,
    thresh=0.05,
    gridsize=200,
    legend=True,
    palette=None,
    color=None,
    linewidth=None,
    linestyle=None,
    alpha=None,
    ax=None
)



where,


| Parameter     | Description                                                        | Short Example                                                    |
| ------------- | ------------------------------------------------------------------ | ---------------------------------------------------------------- |
| `data`        | Dataset (DataFrame, array, list) containing values                 | `sns.kdeplot(data=df)`                                           |
| `x`           | Column name for x-axis (univariate KDE)                            | `sns.kdeplot(data=df, x="Age")`                                  |
| `y`           | Column name for y-axis (bivariate KDE)                             | `sns.kdeplot(data=df, x="Age", y="Salary")`                      |
| `hue`         | Creates separate KDE curves based on categories                    | `sns.kdeplot(data=df, x="Marks", hue="Gender")`                  |
| `fill`        | Fills the area under the KDE curve                                 | `sns.kdeplot(data=df, x="Age", fill=True)`                       |
| `multiple`    | Controls display of multiple KDE curves (`layer`, `stack`, `fill`) | `sns.kdeplot(data=df, x="Age", hue="Group", multiple="stack")`   |
| `bw_adjust`   | Controls smoothing of KDE curve (higher = smoother)                | `sns.kdeplot(data=df, x="Age", bw_adjust=2)`                     |
| `bw_method`   | Bandwidth calculation method (`scott`, `silverman`, value)         | `sns.kdeplot(data=df, x="Age", bw_method="silverman")`           |
| `cut`         | Extends KDE curve beyond minimum and maximum values                | `sns.kdeplot(data=df, x="Age", cut=0)`                           |
| `clip`        | Restricts KDE calculation within a range                           | `sns.kdeplot(data=df, x="Age", clip=(20,60))`                    |
| `cumulative`  | Displays cumulative distribution curve                             | `sns.kdeplot(data=df, x="Age", cumulative=True)`                 |
| `common_norm` | Normalizes KDE curves together when using `hue`                    | `sns.kdeplot(data=df, x="Age", hue="Gender", common_norm=False)` |
| `common_grid` | Uses same evaluation grid for multiple KDE curves                  | `sns.kdeplot(data=df, x="Age", hue="Gender", common_grid=True)`  |
| `levels`      | Number of density contour levels in bivariate KDE                  | `sns.kdeplot(data=df, x="Age", y="Salary", levels=5)`            |
| `thresh`      | Removes contours below a density threshold                         | `sns.kdeplot(data=df, x="Age", y="Salary", thresh=0.1)`          |
| `gridsize`    | Number of grid points used for KDE calculation                     | `sns.kdeplot(data=df, x="Age", gridsize=300)`                    |
| `legend`      | Shows or hides legend                                              | `sns.kdeplot(data=df, x="Age", hue="Gender", legend=False)`      |
| `palette`     | Defines colors for multiple KDE curves                             | `sns.kdeplot(data=df, x="Age", hue="Gender", palette="Set2")`    |
| `color`       | Sets single KDE curve color                                        | `sns.kdeplot(data=df, x="Age", color="red")`                     |
| `linewidth`   | Controls KDE line thickness                                        | `sns.kdeplot(data=df, x="Age", linewidth=3)`                     |
| `linestyle`   | Changes line style                                                 | `sns.kdeplot(data=df, x="Age", linestyle="--")`                  |
| `alpha`       | Controls transparency (0 = transparent, 1 = solid)                 | `sns.kdeplot(data=df, x="Age", fill=True, alpha=0.4)`            |
| `ax`          | Draws plot on a specific Matplotlib axis                           | `sns.kdeplot(data=df, x="Age", ax=ax)`                           |                           
| `log_scale` | Converts x-axis or y-axis into a logarithmic scale. Useful when data values have a very large range. | `sns.kdeplot(data=df, x="Salary", log_scale=True)` |





1.import all required libraries 



```python
import numpy as np
import pandas as pd 
import matplotlib.pyplot as plt
import seaborn as sns 
```

2.import dataset and  specific variables from dataset




```python
df=pd.read_excel("C:/data_analysis_using_python/supermarket_sales.xlsx")
df=df[[ 'Gender','Unit price','Quantity','Total','Payment','gross income']]
df.head()
```

3.Create basic kdeplot

```python
sns.kdeplot(data=df,x="Total")
```

4.Create KDE for all the numeric variable in dataframe 

```python
sns.kdeplot(data=df)
```

5. Adjust the smoothing using bandwidth (bw_adjust)

```python
sns.kdeplot(data=df,x="Total",bw_adjust=0.2)
```

6.Group the KDE on a category variable 

```python
sns.kdeplot(data=df,x="Total",hue="Payment")
```

7.Stack KDE category using MULTIPLE argument 

```python
sns.kdeplot(data=df,x="Total",hue="Payment",multiple="stack")
```

8.Use log scaling to map the variable in KDE 

```python
sns.kdeplot(data=df,x="Total",log_scale=True)
```

9.change styling of hued KDE using  linewidth , palette , alpha . 


```python

sns.kdeplot(data=df,x="Total",hue="Payment",multiple="stack",linewidth=5,palette="Blues",alpha=1)
```

10.Create a bivariate KDE (Two variable plot)

```python
sns.kdeplot(data=df,x="Unit price",y="gross income")
```

11.group the bivariate KDE on categorical variable and show the contours

```python
sns.kdeplot(data=df,x="Unit price",y="gross income",levels=5,thresh=0.2,fill=True)
```
