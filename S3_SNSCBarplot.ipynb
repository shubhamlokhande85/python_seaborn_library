2.Barplot 

A bar plot in Seaborn is a graph used to compare the average (mean) value of a numerical variable for different categories. Each bar represents one category, and its height shows the average value

Syntax 

sns.barplot(
    data=None,
    *,
    x=None,
    y=None,
    hue=None,
    order=None,
    hue_order=None,
    estimator='mean',
    errorbar=('ci', 95),
    n_boot=1000,
    seed=None,
    orient=None,
    color=None,
    palette=None,
    saturation=0.75,
    width=0.8,
    dodge='auto',
    gap=0,
    capsize=0,
    err_kws=None,
    ci='deprecated',
    errcolor='deprecated',
    errwidth='deprecated',
    ax=None,
    **kwargs
)

where,

| **Parameter**  | **Purpose**                                                                                                              | **Example**                             |
| -------------- | ------------------------------------------------------------------------------------------------------------------------ | --------------------------------------- |
| **data**       | Specifies the DataFrame that contains the dataset used to create the bar plot.                                           | `data=tips`                             |
| **x**          | Selects the categorical column to display on the x-axis.                                                                 | `x="day"`                               |
| **y**          | Selects the numerical column whose values are summarized (mean by default).                                              | `y="total_bill"`                        |
| **hue**        | Adds another categorical variable to create grouped bars with different colors.                                          | `hue="sex"`                             |
| **order**      | Defines the custom order of categories on the x-axis.                                                                    | `order=["Sun","Sat","Fri","Thur"]`      |
| **hue_order**  | Defines the custom order of hue categories.                                                                              | `hue_order=["Male","Female"]`           |
| **estimator**  | Function used to calculate the height of each bar. Default is **mean**. Other options include **median, sum, max, min**. | `estimator=np.median`                   |
| **errorbar**   | Displays uncertainty using confidence intervals, standard deviation, or standard error.                                  | `errorbar="sd"`                         |
| **n_boot**     | Number of bootstrap samples used to estimate confidence intervals. Default is **1000**.                                  | `n_boot=500`                            |
| **seed**       | Sets a random seed so the results remain the same each time the code is run.                                             | `seed=42`                               |
| **orient**     | Sets the orientation of the plot (`"v"` for vertical, `"h"` for horizontal).                                             | `orient="h"`                            |
| **color**      | Assigns a single color to all bars.                                                                                      | `color="green"`                         |
| **palette**    | Applies multiple colors using a predefined color palette.                                                                | `palette="viridis"`                     |
| **saturation** | Controls the intensity of bar colors (0 = dull, 1 = full color).                                                         | `saturation=0.9`                        |
| **width**      | Adjusts the width of each bar. Default is **0.8**.                                                                       | `width=0.5`                             |
| **dodge**      | Separates grouped bars when `hue` is used.                                                                               | `dodge=True`                            |
| **gap**        | Adds spacing between grouped bars.                                                                                       | `gap=0.2`                               |
| **capsize**    | Sets the size of the caps on the error bars.                                                                             | `capsize=0.3`                           |
| **err_kws**    | Customizes the appearance of error bars (color, linewidth, etc.).                                                        | `err_kws={"color":"red","linewidth":2}` |
| **ax**         | Specifies the Matplotlib Axes object on which the plot will be drawn.                                                    | `ax=ax1`                                |
| **kwargs**     | Additional arguments passed to Matplotlib's `bar()` function.                                                            | `edgecolor="black"`                     |
   

Quick Tips 


| **Parameter Type** | **Parameters**                                                         |
| ------------------ | ---------------------------------------------------------------------- |
| **Data Selection** | `data`, `x`, `y`, `hue`                                                |
| **Category Order** | `order`, `hue_order`                                                   |
| **Calculation**    | `estimator`, `errorbar`, `n_boot`, `seed`                              |
| **Appearance**     | `color`, `palette`, `saturation`, `width`, `gap`, `capsize`, `err_kws` |
| **Layout**         | `orient`, `dodge`, `ax`, `kwargs`                                      |

Important Concepts 

Errorbar 

The errorbar parameter is used to display the uncertainty or variation in the data by drawing error bars on top of each bar in a bar plot.


Error bars help you understand how much the data varies around the estimated value (such as the mean)


syntax :


errorbar=values 

Common Values 

| Value  | Description                                          |
| ------ | ---------------------------------------------------- |
| `"ci"` | Displays a **confidence interval** (default is 95%). |
| `"sd"` | Displays the **standard deviation** of the data.     |
| `"se"` | Displays the **standard error** of the mean.         |
| `None` | Removes error bars from the plot.                    |
