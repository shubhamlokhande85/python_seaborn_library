6.Swarm plot


A Swarm Plot is a Seaborn plot that shows each data point separately and automatically adjusts the position of the dots so that they do not overlap.


Syntax


sns.swarmplot(x=None, y=None, hue=None, data=None,
              order=None, hue_order=None,
              dodge=False, orient=None,
              color=None, palette=None,
              size=5, edgecolor="gray",
              linewidth=0)


where, 


| Parameter   | Description                                                  | Example                            |
| ----------- | ------------------------------------------------------------ | ---------------------------------- |
| `x`         | Specifies the categorical variable on the X-axis.            | `x="day"`                          |
| `y`         | Specifies the numerical variable on the Y-axis.              | `y="total_bill"`                   |
| `hue`       | Groups data into different colors based on another category. | `hue="sex"`                        |
| `data`      | Dataset containing the variables.                            | `data=tips`                        |
| `order`     | Specifies the order of categories on the X-axis.             | `order=["Thur","Fri","Sat","Sun"]` |
| `hue_order` | Specifies the order of hue categories.                       | `hue_order=["Male","Female"]`      |
| `dodge`     | Separates hue groups so their dots don't overlap.            | `dodge=True`                       |
| `orient`    | Sets the orientation. `"v"` = vertical, `"h"` = horizontal.  | `orient="h"`                       |
| `color`     | Sets one color for all dots.                                 | `color="red"`                      |
| `palette`   | Assigns different colors to hue groups.                      | `palette="Set2"`                   |
| `size`      | Changes the size of the dots.                                | `size=8`                           |
| `edgecolor` | Sets the border color of dots.                               | `edgecolor="black"`                |
| `linewidth` | Sets the thickness of the dot border.                        | `linewidth=1`                      |



Difference Between Strip Plot and Swarm Plot



| Strip Plot                                        | Swarm Plot                                                      |
| ------------------------------------------------- | --------------------------------------------------------------- |
| Dots may overlap, so `jitter=True` is often used. | Dots never overlap because Seaborn arranges them automatically. |
| Faster for large datasets.                        | Better for small to medium datasets with clear visibility.      |
| Uses the `jitter` parameter.                      | Does **not** use `jitter`.                                      |

1.Import all required libraries 

```python
import numpy as np
import pandas as pd 
import matplotlib.pyplot  as plt
import seaborn as sns
```

2.Import dataset


```python
df=pd.read_excel("C:/data_analysis_using_python/supermarket_sales.xlsx")
df.head()
```

3.Create a basic swarmplot

```python
sns.swarmplot(data=df,y="Total")
```

4.Create swarmplot group by categories 

```python
sns.swarmplot(data=df,y="Total",x="Payment",hue="Gender")
```

5.Showing hues separatly on the categorical axis

```python
sns.swarmplot(data=df,y="Total",x="Payment",hue="Gender", dodge=True)
```

6.Styling  a swarm - change the marker , size, color , edgecolor

```python
plt.figure(figsize=(15,5))
sns.swarmplot(data=df,y="Total",x="Payment",marker="*",size=7,color='Green',edgecolor="red")
```

7.Overlay swarmplot on a boxplot

```python
sns.boxplot(data=df,y="Total",x="Payment",color='Red')
sns.swarmplot(data=df,y="Total",x="Payment",color='black')
```

8.Overlay swarmplot on a violinplot

```python
plt.figure(figsize=(15,5))
sns.violinplot(data=df,y="Total",x="Payment",color="red")
sns.swarmplot(data=df,y="Total",x="Payment",color="black")
```
