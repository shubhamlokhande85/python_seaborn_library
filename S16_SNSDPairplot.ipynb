7.Pairplot 


A pairplot in Seaborn is a visualization used to compare multiple numerical variables with each other. It creates a grid of plots:

-Scatter plots  - show the relationship between two different variables.


- intersection of two different variables 

-Distribution plots - show the distribution of each individual variable.



 - intersection of same variable


It is useful for finding correlation, patterns, trends, clusters, and outliers



-syntax 


sns.pairplot(
    data,
    hue=None,
    vars=None,
    x_vars=None,
    y_vars=None,
    kind='scatter',
    diag_kind='auto',
    markers=None,
    palette=None,
    corner=False
)


where,


| Parameter     | Meaning                                                       | Example                                           |
| ------------- | ------------------------------------------------------------- | ------------------------------------------------- |
| `data`        | DataFrame containing the variables                            | `sns.pairplot(df)`                                |
| `hue`         | Colors the plots according to a categorical column            | `sns.pairplot(df, hue="Gender")`                  |
| `vars`        | Selects specific columns for the pairplot                     | `sns.pairplot(df, vars=["Math", "Science"])`      |
| `x_vars`      | Variables to display on the X-axis                            | `sns.pairplot(df, x_vars=["Math", "Science"])`    |
| `y_vars`      | Variables to display on the Y-axis                            | `sns.pairplot(df, y_vars=["English", "Science"])` |
| `kind`        | Type of plot for relationships, such as scatter or regression | `sns.pairplot(df, kind="reg")`                    |
| `diag_kind`   | Plot used on the diagonal, such as histogram or KDE           | `sns.pairplot(df, diag_kind="kde")`               |
| `markers`     | Changes the shape of data points                              | `sns.pairplot(df, markers="o")`                   |
| `palette`     | Sets the colors used for different categories                 | `sns.pairplot(df, hue="Gender", palette="Set1")`  |
| `corner=True` | Shows only the lower triangle and removes duplicate plots     | `sns.pairplot(df, corner=True)`                   |



-Diagonal


In a pairplot, the diagonal is the set of plots where a variable is compared with itself.


-Eg



-For example, if we have:

English

```python
#Eg .2 specific x axis variables instersection with specific y axis variables 
sns.pairplot(data=df,x_vars=["sepal_length","petal_length","petal_width"],
             y_vars=["sepal_width","sepal_length"],diag_kind=None)
```

8.showing only the lower triangle using cornor argument

```python
sns.pairplot(data=df,corner=True)
```

9.making changes specific to diagonal

```python
sns.pairplot(data=df,diag_kws=dict(color="red",kde=True))
```

10 . Making changes specific to non diagonal plots 

```python
sns.pairplot(data=df,plot_kws=dict(color="green",marker=10,s=50))
'''
marker = 10 
s = 50

 
where,
10 -  is denotes pyramid 
s  - is denotes size of marker 
'''
```

11. KDE plot on Top of the pair plot using map_lower or map_upper

```python
i =sns.pairplot(data=df,plot_kws=dict(color="red"))
i.map_upper(sns.kdeplot)
i.map_lower(sns.kdeplot)
```

When we create a pair plot/grid, we can control what appears in the upper triangle and lower triangle.



.map_upper() → applies a plotting function to the upper triangle of the grid.


.map_lower() → applies a plotting function to the lower triangle of the grid.

Think of the grid like this:

Height	      Weight	       Age


Height	   —	     🟦 Upper	🟦 Upper



Weight	   🟥 Lower	 —	        🟦 Upper



Age....  .           🟥 Lower	   🟥 Lower	    —




🟦 = map_upper()


🟥 = map_lower()



syntax 
plot.map_upper(function)
plot.map_lower(function)

function - sns.plot_names
