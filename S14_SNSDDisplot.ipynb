B.Distribution plots


5.Displot


displot() helps us understand how values are spread in a dataset.


displot() is a figure-level function in Seaborn used to visualize the distribution of one or more variables. It can create histograms, KDE plots, ECDF plots, and combinations of these



Syntax


sns.displot(
    data=None,
    x=None,
    y=None,
    hue=None,
    row=None,
    col=None,
    kind="hist",
    bins=None,
    kde=False,
    rug=False,
    stat="count",
    multiple="layer",
    palette=None,
    height=5,
    aspect=1
)

where,



| Parameter  | Simple meaning                            | Example                                                      |
| ---------- | ----------------------------------------- | ------------------------------------------------------------ |
| `data`     | Dataset/DataFrame to use                  | `sns.displot(data=df, x="marks")`                            |
| `x`        | Column shown on X-axis                    | `sns.displot(df, x="marks")`                                 |
| `y`        | Column shown on Y-axis                    | `sns.displot(df, x="marks", y="age")`                        |
| `hue`      | Divides data into different groups/colors | `sns.displot(df, x="marks", hue="gender")`                   |
| `kind`     | Type of distribution plot                 | `sns.displot(df, x="marks", kind="kde")`                     |
| `bins`     | Number of bars in histogram               | `sns.displot(df, x="marks", bins=10)`                        |
| `kde`      | Adds a smooth distribution curve          | `sns.displot(df, x="marks", kde=True)`                       |
| `rug`      | Shows individual data points              | `sns.displot(df, x="marks", rug=True)`                       |
| `row`      | Creates plots row-wise                    | `sns.displot(df, x="marks", row="gender")`                   |
| `col`      | Creates plots column-wise                 | `sns.displot(df, x="marks", col="class")`                    |
| `stat`     | Decides what histogram measures           | `sns.displot(df, x="marks", stat="density")`                 |
| `multiple` | Controls how groups are displayed         | `sns.displot(df, x="marks", hue="gender", multiple="stack")` |
| `palette`  | Sets colors                               | `sns.displot(df, x="marks", hue="gender", palette="Set2")`   |
| `height`   | Controls plot height                      | `sns.displot(df, x="marks", height=6)`                       |
| `aspect`   | Controls plot width                       | `sns.displot(df, x="marks", aspect=1.5)`                     |


1.Import all required libraries 

```python
import numpy as np
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns
```

2.import dataset 

```python
df=pd.read_excel("C:/data_analysis_using_python/supermarket_sales.xlsx")
df=df[['Branch','Gender','Quantity','Total','Payment','gross income']]
df.head()
```

3.Create a basic displot

```python
sns.displot(data=df,x="Total",kde=True,rug=True,rug_kws=dict(height=0.05)) 
# To rug_kws+ you can pass attributes like this rug_kws={'height=0.05}
```

4.Create categories displot using hue 

```python
sns.displot(data=df,x="Total",kde=True,rug=True,hue="Payment",multiple='stack')
```

5.Visualize displot in further categories 

```python
sns.displot(data=df,x='Total',hue="Payment",kde=True, rug=True,col="Branch",row='Gender')
```

6.Create ECDFplot and visualize in further categories

```python
sns.displot(data=df,x="Total",hue="Payment",kind='ecdf',col="Branch",row="Gender")
```
