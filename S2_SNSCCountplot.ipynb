A.Categorial plots 


Categorical plots in Seaborn are visualization tools used to display, compare, and analyze data based on different categories or group

1.Countplot



A countplot in Seaborn is a type of graph used to show the number of observations (frequency/count) 
 for each category in a dataset. It is mainly used for categorical data and displays the count of each
 category using bars


 -Syntax

 sns.countplot(data, x, y, hue, palette, order)

 
 where,

 
| Parameter   | Meaning                                                                  |
| ----------- | ------------------------------------------------------------------------ |
| **data**    | The dataset (DataFrame) containing the values                            |
| **x**       | Column name to display on the x-axis                                     |
| **y**       | Column name to display on the y-axis (instead of x)                      |
| **hue**     | Used to divide data into different groups using colors                   |
| **palette** | Sets the colors of the bars                                              |
| **order**   | Controls the order of categories displayed                               |
| **orient**  | Changes the direction of the plot (`v` for vertical, `h` for horizontal) |








Example

1.Importing all required libraries 

```python
import numpy as np
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns 
```

2.Import the data

```python
df=pd.read_excel("C:/data_analysis_using_python/supermarket_sales.xlsx")
df.head()
```

3.Create a basic countplot to get number of transaction for each of product line and change figure size

```python
plt.figure(figsize=(15,5))
sns.countplot(x="Product line", data=df)
```

NOTE


plt.figure()


plt.figure() is a Matplotlib function in Python that creates a new blank drawing area (figure/canvas) where we can make graphs and charts. It is used to control the size, color, and quality of the graph before plotting data


Syntax


matplotlib.pyplot.figure(
    num=None,
    figsize=None,
    dpi=None,
    facecolor=None,
    edgecolor=None,
    frameon=True,
    FigureClass=Figure,
    clear=False,
    **kwargs
)


where, 

| Parameter     | Description                                                                    |
| ------------- | ------------------------------------------------------------------------------ |
| `num`         | Optional figure number or name. Used to identify a specific figure.            |
| `figsize`     | Tuple specifying figure size as `(width, height)` in inches. Example: `(8, 5)` |
| `dpi`         | Dots per inch (resolution of the figure). Higher DPI gives better quality.     |
| `facecolor`   | Background color of the figure. Example: `"white"`, `"lightgray"`              |
| `edgecolor`   | Color of the figure border.                                                    |
| `frameon`     | If `True`, displays the figure frame; if `False`, hides it.                    |
| `FigureClass` | Custom Figure class to create instead of the default Matplotlib Figure.        |
| `clear`       | If `True`, clears the existing figure with the same number.                    |
| `**kwargs`    | Additional keyword arguments passed to the Figure object.                      |


4.Make it horizontal bar plot 

```python
plt.figure(figsize=(15,5))
sns.countplot(y="Product line",data=df)
```

5.Add hue to get count on two categories i.e product line and gender 

```python
plt.figure(figsize=(15,5))
sns.countplot(x="Product line", hue="Gender", data=df)
```

6.Using different colour palette 

```python
plt.figure(figsize=(15,5))
sns.countplot(x="Product line",hue="Gender", data=df, palette="Reds")
```

7.Change style using face color , line width and edge color 

```python
plt.figure(figsize=(15,5))
sns.countplot(x="Product line",data=df ,
              facecolor=(0,0,0,0),
              linewidth=5,
              edgecolor=sns.color_palette('dark',3))
```
