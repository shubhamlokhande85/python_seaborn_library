3.Boxplot


A Box Plot is a statistical graph used to display the distribution of numerical data across different categories. It summarizes the data using the minimum value, first quartile (Q1), median (Q2), third quartile (Q3), maximum value, and outliers


Syntax 


sns.boxplot(
    x=None,
    y=None,
    hue=None,
    data=None,
    order=None,
    hue_order=None,
    orient=None,
    color=None,
    palette=None,
    width=0.8,
    dodge=True,
    whis=1.5,
    linewidth=None
) 

where,


| Parameter   | Description                                                         |
| ----------- | ------------------------------------------------------------------- |
| `x`         | Categorical variable placed on the x-axis.                          |
| `y`         | Numerical variable placed on the y-axis.                            |
| `hue`       | Creates separate box plots based on another categorical variable.   |
| `data`      | Dataset containing the variables.                                   |
| `order`     | Specifies the order of categories on the axis.                      |
| `hue_order` | Specifies the order of hue categories.                              |
| `orient`    | Orientation of the plot (`'v'` for vertical, `'h'` for horizontal). |
| `color`     | Sets a single color for all box plots.                              |
| `palette`   | Applies multiple colors to different categories.                    |
| `width`     | Controls the width of the boxes (default `0.8`).                    |
| `dodge`     | Separates boxes when `hue` is used (`True` or `False`).             |
| `whis`      | Controls the length of the whiskers. Default is `1.5 × IQR`.        |
| `linewidth` | Sets the thickness of the box borders.                              |



NOTE : 


1. Categorical variable

A categorical variable is a variable that represents groups, labels, or categories instead of numbers (Text rows)

Eg.


Gender (Male, Female),
Color (Red, Blue, Green),
Department (HR, Sales, IT)




2.Numerical variable 

A numerical variable is a variable that represents numbers or measurable values (Numerical rows)

Eg.

Age (18, 25, 30),
Height (160 cm, 170 cm),
Salary (₹25,000, ₹40,000)

1. Import all required libraries 

```python
import numpy as np
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns 
```

2.Import data

```python
df=pd.read_excel("C:/data_analysis_using_python/supermarket_sales.xlsx")
df.head()
```

3.Create basic boxplot on one numeric variable 

```python
sns.boxplot(y='Total',data=df)
```

4.Create a basic boxplot on one numeric variable by a categorical variable

```python
# Classifying the payment method(categorical variable) using one numeric variable (Total)
sns.boxplot(x='Payment',y='Total',data=df)
```

5.Create a basic boxplot on one numeric variable by two categorical variable using hue attribute

```python
# dividing payment method male and female 
sns.boxplot(x='Payment',y='Total',hue='Gender',data=df)
```

6.Add MEAN marker in the box plot using showmean attribute and change its style meanprops

```python
# Displaying Mean and its properties 
sns.boxplot(x='Payment',y='Total',hue='Gender',data=df,showmeans=True,meanprops={"marker":"o"
                                                                        ,"markerfacecolor":"white"
                                                                        ,"markersize":"10"
                                                                        ,"markeredgecolor":"black"})
```

7.Make boxplot horizontal

```python
# just change axis's values 
sns.boxplot(y='Payment',x='Total',hue='Gender',data=df,showmeans=True,meanprops={"marker":"o"
                                                                        ,"markerfacecolor":"white"
                                                                        ,"markersize":"10"
                                                                        ,"markeredgecolor":"black"}) 
```

8.chane color boxplot and its width of line using palette and linewidth 

```python
sns.boxplot(y='Payment',x='Total',hue='Gender',data=df,showmeans=True,meanprops={"marker":"o"
                                                                        ,"markerfacecolor":"white"
                                                                        ,"markersize":"10"
                                                                        ,"markeredgecolor":"black"}
            ,palette='Greens',linewidth=2) 
```
