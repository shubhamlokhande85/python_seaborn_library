A.Categorical plots

7.Catplot/ factorplot


seaborn.catplot() is a figure-level plotting function that combines categorical plotting functions with the FacetGrid interface, allowing you to create statistical graphics and easily split them across rows, columns, or categories


Syntax


import seaborn as sns

sns.catplot(
    data=None,
    x=None,
    y=None,
    hue=None,
    row=None,
    col=None,
    kind="strip",
    estimator="mean",
    errorbar=("ci", 95),
    n_boot=1000,
    seed=None,
    units=None,
    weights=None,
    order=None,
    hue_order=None,
    row_order=None,
    col_order=None,
    col_wrap=None,
    height=5,
    aspect=1,
    orient=None,
    color=None,
    palette=None,
    legend="auto",
    legend_out=True,
    sharex=True,
    sharey=True,
    margin_titles=False,
    facet_kws=None,
    native_scale=False,
    formatter=None,
    **kwargs
)


where, 



| Parameter         | Short Explanation                                    | Example                      |
| ----------------- | ---------------------------------------------------- | ---------------------------- |
| **data**          | DataFrame containing the dataset                     | `data=tips`                  |
| **x**             | Categorical variable on X-axis                       | `x="day"`                    |
| **y**             | Numeric variable on Y-axis                           | `y="total_bill"`             |
| **hue**           | Adds another category using different colors         | `hue="sex"`                  |
| **row**           | Creates separate rows of plots                       | `row="smoker"`               |
| **col**           | Creates separate columns of plots                    | `col="time"`                 |
| **kind**          | Chooses the chart type                               | `"box"`, `"bar"`, `"violin"` |
| **estimator**     | Function used to summarize data (bar/point plots)    | `np.mean`, `np.median`       |
| **errorbar**      | Displays error bars                                  | `("ci",95)`, `None`          |
| **n_boot**        | Number of bootstrap samples for confidence intervals | `1000`                       |
| **seed**          | Random seed for reproducible results                 | `seed=42`                    |
| **units**         | Identifies repeated measurements                     | `units="id"`                 |
| **weights**       | Weights for weighted statistics                      | `weights="weight"`           |
| **order**         | Sets category order on x-axis                        | `["Sun","Sat","Fri"]`        |
| **hue_order**     | Sets hue category order                              | `["Female","Male"]`          |
| **row_order**     | Order of row facets                                  | `["Yes","No"]`               |
| **col_order**     | Order of column facets                               | `["Lunch","Dinner"]`         |
| **col_wrap**      | Wraps facet columns after a fixed number             | `col_wrap=2`                 |
| **height**        | Height of each subplot (inches)                      | `height=5`                   |
| **aspect**        | Width = height × aspect                              | `aspect=1.5`                 |
| **orient**        | Plot orientation                                     | `"v"` or `"h"`               |
| **color**         | Single color for the plot                            | `"red"`                      |
| **palette**       | Color palette                                        | `"Set2"`, `"deep"`           |
| **legend**        | Controls legend display                              | `"auto"`, `False`            |
| **legend_out**    | Places legend outside the plot                       | `True`                       |
| **sharex**        | Share x-axis across facets                           | `True`                       |
| **sharey**        | Share y-axis across facets                           | `False`                      |
| **margin_titles** | Displays facet titles in margins                     | `True`                       |
| **facet_kws**     | Extra options for `FacetGrid`                        | `{"despine": False}`         |
| **native_scale**  | Uses original numeric/datetime scale                 | `True`                       |
| **formatter**     | Formats category labels                              | `str.upper`                  |
| **`**kwargs`**    | Extra arguments passed to the selected plot type     | `linewidth=2`, `fill=False`  |


Example


1.import all required libraries 

```python
import numpy as np
import pandas as pd 
import matplotlib.pyplot
import seaborn as sns 
```

2.import dataset

```python
df=pd.read_excel("C:/data_analysis_using_python/supermarket_sales.xlsx")
df.head()
```

3.Create a catplot

```python
sns.catplot(data=df,y="Total",x="Payment")
```

4.Create a catplot - Barplot

```python
#To create different types of chart we use kind argument 
sns.catplot(data=df , y="Total",x="Payment",kind='bar')
```

5.Create violin plot using catplot and define payment method by male and female individually and also as well as customer type


```python
sns.catplot(data=df,y="Total",x="Payment",col="Gender",row="Customer type",hue="City",kind="violin")
```
