C.Relational plots

3.Relplot 

A relational plot is used to show the relationship between two numerical variables.


relplot() is a figure-level function in Seaborn. It provides a convenient way to create relational plots and can create either scatter plots or line plots using the kind parameter.


By default, relplot() creates a scatter plot (kind="scatter").

It is especially useful when you want to visualize relationships across multiple categories or subplots using row and col.


Syntax 


sns.relplot(
    data=None,
    *,
    x=None,
    y=None,
    hue=None,
    size=None,
    style=None,
    units=None,
    weights=None,
    row=None,
    col=None,
    col_wrap=None,
    row_order=None,
    col_order=None,
    palette=None,
    hue_order=None,
    hue_norm=None,
    sizes=None,
    size_order=None,
    size_norm=None,
    markers=None,
    dashes=None,
    style_order=None,
    legend="auto",
    kind="scatter",
    height=5,
    aspect=1,
    facet_kws=None,
    units=None,
    weights=None,
    estimator="mean",
    errorbar=("ci", 95),
    n_boot=1000,
    seed=None,
    sort=True,
    err_style="band",
    err_kws=None,
    legend="auto",
    height=5,
    aspect=1,
    facet_kws=None,
    **kwargs
)


where, 

| Parameter     | Description                                                                                             | Example                            |
| ------------- | ------------------------------------------------------------------------------------------------------- | ---------------------------------- |
| `data`        | DataFrame containing the required columns.                                                              | `data=df`                          |
| `x`           | Column used for the **X-axis**.                                                                         | `x="Month"`                        |
| `y`           | Column used for the **Y-axis**.                                                                         | `y="Sales"`                        |
| `hue`         | Creates different colors based on a variable/category.                                                  | `hue="Gender"`                     |
| `size`        | Changes the size/width of points or lines based on a variable.                                          | `size="Category"`                  |
| `style`       | Gives different marker or line styles based on a variable.                                              | `style="Gender"`                   |
| `units`       | Draws a separate line for each observation/group without aggregating. Mainly useful with `kind="line"`. | `units="Student_ID"`               |
| `weights`     | Column containing weights used in statistical estimation.                                               | `weights="Frequency"`              |
| `row`         | Creates separate subplots for different categories along rows.                                          | `row="Gender"`                     |
| `col`         | Creates separate subplots for different categories along columns.                                       | `col="Year"`                       |
| `col_wrap`    | Wraps column facets into multiple rows after the specified number of columns.                           | `col_wrap=3`                       |
| `row_order`   | Specifies the order of categories in `row`.                                                             | `row_order=["Male","Female"]`      |
| `col_order`   | Specifies the order of categories in `col`.                                                             | `col_order=[2024,2025]`            |
| `palette`     | Specifies colors used for different groups.                                                             | `palette="Set2"`                   |
| `hue_order`   | Specifies the order of categories displayed by `hue`.                                                   | `hue_order=["Male","Female"]`      |
| `hue_norm`    | Normalizes numeric values used with `hue`.                                                              | `hue_norm=(0,100)`                 |
| `sizes`       | Specifies the range or mapping of sizes.                                                                | `sizes=(20,200)`                   |
| `size_order`  | Specifies the order of categories displayed by `size`.                                                  | `size_order=["Small","Large"]`     |
| `size_norm`   | Normalizes numeric values used with `size`.                                                             | `size_norm=(0,100)`                |
| `markers`     | Controls the markers used for different groups.                                                         | `markers=True`                     |
| `dashes`      | Controls dashed/solid line styles. Mainly used with `kind="line"`.                                      | `dashes=True`                      |
| `style_order` | Specifies the order of categories displayed by `style`.                                                 | `style_order=["A","B"]`            |
| `kind`        | Specifies the type of relational plot.                                                                  | `kind="scatter"` or `kind="line"`  |
| `legend`      | Controls whether and how the legend is displayed.                                                       | `legend="auto"`                    |
| `height`      | Height of each facet/subplot in inches.                                                                 | `height=5`                         |
| `aspect`      | Ratio of width to height of each facet.                                                                 | `aspect=1.5`                       |
| `facet_kws`   | Dictionary of additional arguments passed to the FacetGrid.                                             | `facet_kws={"margin_titles":True}` |
| `estimator`   | Statistical function used when multiple observations have the same `x` value. Mainly for line plots.    | `estimator="mean"`                 |
| `errorbar`    | Controls the error interval around the estimated line.                                                  | `errorbar=("ci",95)`               |
| `n_boot`      | Number of bootstrap iterations used to estimate confidence intervals.                                   | `n_boot=1000`                      |
| `seed`        | Seed used for reproducible bootstrap calculations.                                                      | `seed=42`                          |
| `sort`        | Sorts data according to the X-axis before drawing lines.                                                | `sort=True`                        |
| `err_style`   | Controls how the error interval is displayed.                                                           | `err_style="band"`                 |
| `err_kws`     | Additional keyword arguments for the error display.                                                     | `err_kws={"alpha":0.2}`            |


1.Import all required libraries 

```python
import numpy as np
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns 
```

2.Import dataset 

```python
df=pd.read_excel("C:/Python_Seaborn_Library/mart_lineplot.xlsx").sample(n=50)

#Making column names in small letter 
df.columns=df.columns.str.lower()

#first five rows of dataset 
df.head()
```

3.create basic relplot 

```python
sns.relplot(data=df,x="outlet_year",y="sales")
```

4.Create relplot using kind parameter 


```python
sns.relplot(data=df, x="outlet_year",y="sales",kind="line")
```

5.Create a relplot based of categorical variable 

```python
sns.relplot(data=df,x="outlet_year",y="sales",hue="outlet_size",s=100)
```

6.Create relplot based on categorical variable's unique items when they are limited or measurable 

```python
sns.relplot(data=df,x="outlet_year",y="sales",s=100
            ,hue='outlet_size'
            ,col='tier')
```

7.Create relplot based on categorical variable's unique items when they are unlimited or more

```python
# case 01 
sns.relplot(data=df,x="outlet_year",y="sales",s=100
           ,hue="outlet_size"
           ,col="item_type")
```

```python
# in upper chart we can't see what kind of relationship have between two variables so uppper chart not make a sense 
# to solve this probleum we col_wrap parameter 
#case 02
sns.relplot(data=df,x="outlet_year",y="sales",s=100
           ,hue="outlet_size"
           ,col="item_type"
           ,col_wrap=5)
```

8.Create relplot using two categorical variable 

```python
sns.relplot(data=df,x='outlet_year',y='sales',s=100
            ,hue='outlet_size'
            ,col='tier'
            ,row='outlet_size')
```

9.style the relplot

```python

color=['green','red','blue']
sns.relplot(data=df, x="outlet_year",y="sales",s=100
            ,hue='outlet_size'
            ,palette=color)
```
