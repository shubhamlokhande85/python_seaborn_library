E. MATRIX PLOT 


A matrix plot in Seaborn is a type of visualization that displays data in a grid (rows and columns), where each cell represents a value.

The color of each cell can be used to show the size or strength of the value, making it easy to compare many values at once


1.Heatmaps 


A heatmap in Seaborn is a graphical representation of data in the form of a matrix, where different colors represent different values.

Syntax 

sns.heatmap(
    data,
    *,
    vmin=None,
    vmax=None,
    cmap=None,
    center=None,
    robust=False,
    annot=None,
    fmt=".2g",
    annot_kws=None,
    linewidths=0,
    linecolor="white",
    cbar=True,
    cbar_kws=None,
    cbar_ax=None,
    square=False,
    xticklabels="auto",
    yticklabels="auto",
    mask=None,
    ax=None,
    **kwargs
)

where , 

| Parameter     | Description                                                              | Example                            |
| ------------- | ------------------------------------------------------------------------ | ---------------------------------- |
| `data`        | Matrix/DataFrame that you want to display                                | `data=df`                          |
| `vmin`        | Sets the minimum value for the color scale                               | `vmin=-1`                          |
| `vmax`        | Sets the maximum value for the color scale                               | `vmax=1`                           |
| `cmap`        | Sets the color scheme of the heatmap                                     | `cmap="coolwarm"`                  |
| `center`      | Sets the value at the center of the color map                            | `center=0`                         |
| `robust`      | Uses robust values for calculating the color range; useful with outliers | `robust=True`                      |
| `annot`       | Displays actual values inside cells                                      | `annot=True`                       |
| `fmt`         | Controls the number format shown in cells                                | `fmt=".2f"`                        |
| `annot_kws`   | Customizes annotation text                                               | `annot_kws={"size":12}`            |
| `linewidths`  | Sets the width of lines between cells                                    | `linewidths=0.5`                   |
| `linecolor`   | Sets the color of lines between cells                                    | `linecolor="white"`                |
| `cbar`        | Shows or hides the color bar                                             | `cbar=True`                        |
| `cbar_kws`    | Customizes the color bar                                                 | `cbar_kws={"label":"Correlation"}` |
| `square`      | Makes each cell square-shaped                                            | `square=True`                      |
| `xticklabels` | Controls labels on the X-axis                                            | `xticklabels=True`                 |
| `yticklabels` | Controls labels on the Y-axis                                            | `yticklabels=True`                 |
| `mask`        | Hides selected cells                                                     | `mask=mask`                        |
| `ax`          | Specifies the Matplotlib axes                                            | `ax=ax`                            |


NOTE : 


Before creating a matrix plot, we prepare the data in a matrix (2D tabular) format, such as a pivot table or correlation matrix, so that each cell contains a numerical value that can be represented using colors


| Method                 | Purpose                                         | Example                |
| ---------------------- | ----------------------------------------------- | ---------------------- |
| **Pivot table**        | Summarizes data by rows and columns             | `df.pivot_table(...)`  |
| **Correlation matrix** | Shows relationships between numerical variables | `df.corr()`            |
| **Crosstab**           | Creates a frequency/count matrix                | `pd.crosstab(...)`     |
| **Custom matrix**      | Manually created 2D numerical data              | `data = [[1,2],[3,4]]` |




1. import all required libraries 

```python
import numpy as np
import pandas as pd 
import matplotlib.pyplot as plt
import seaborn as sns
```

2.import data

```python
df=pd.read_excel("C:/Python_Seaborn_Library/mart_linePlot.xlsx")
df.columns = df.columns.str.lower()
df.head(10)
```

3.shape of data

```python
df.shape
```

4. Convert data into matrix form of data using pivot_table()

```python
df.pivot_table(index='outlet_year', columns="outlet_size",values="sales")
```

4. Store matrix form of data in a variable 

```python
dfpiv=df.pivot_table(index='outlet_year', columns="outlet_size",values="sales")
```

5.Create a heatmap 

```python
sns.heatmap(dfpiv)
```

6.Styling and highlighting the data in the heatmap

```python
sns.heatmap(dfpiv
            ,annot=True
            ,fmt=".0f"
            ,annot_kws=dict(size=15, weight='bold')
            ,linewidth=0.5
            ,linecolor="black"
            ,cmap='Reds'
            # ,square= True
            )
```

7. Convert pivot data into matrix form data and using correlation 

```python
dfpiv.corr()
```

8.create correlational heatmap 

```python
dfcor=dfpiv.corr()
sns.heatmap(dfcor)
```

9.styling and arrangement in correlational heatmap 

```python
sns.heatmap(dfcor
            ,vmin=1
            ,vmax=1
            # ,center=0
            ,cmap="Reds"
            ,annot=True
            ,annot_kws=dict(size=15, weight="bold")
            ,linewidth=0.5
            ,linecolor="black")
```
