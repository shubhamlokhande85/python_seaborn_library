1.subplot


subplot() is a Matplotlib function used to create multiple plots (charts) inside one figure.

It divides one figure area into a grid of rows and columns and allows each chart to be placed in a specific position.


Syntax


import matplotlib.pyplot as plt
import seaborn as sns

fig, ax = plt.subplots(
    nrows=1,
    ncols=1,
    figsize=None,
    dpi=None,
    sharex=False,
    sharey=False,
    squeeze=True,
    subplot_kw=None,
    gridspec_kw=None
)


where,

| Parameter     | Short Meaning                | Example                           |
| ------------- | ---------------------------- | --------------------------------- |
| `fig`         | Complete figure/canvas       | Whole dashboard                   |
| `ax`          | Individual chart area        | Where plot is drawn               |
| `nrows`       | Number of rows               | `nrows=2` → 2 charts vertically   |
| `ncols`       | Number of columns            | `ncols=3` → 3 charts horizontally |
| `figsize`     | Figure size `(width,height)` | `figsize=(10,5)`                  |
| `dpi`         | Image quality                | `dpi=150`                         |
| `sharex`      | Same X-axis for all charts   | `sharex=True`                     |
| `sharey`      | Same Y-axis for all charts   | `sharey=True`                     |
| `squeeze`     | Controls `ax` format         | Single axis or array              |
| `subplot_kw`  | Extra chart settings         | `projection="polar"`              |
| `gridspec_kw` | Controls spacing/layout      | `hspace`, `wspace`                |






1.Import all required libraries


```python
import numpy as np
import pandas as pd 
import matplotlib.pyplot as plt
import seaborn as sns 
```

2.Import dataset

```python
df=pd.read_excel("C:/data_analysis_using_python/supermarket_sales.xlsx")
df.head()
```

3.Creating  dashboard 

```python
fig, ax = plt.subplots(
    nrows=2,
    ncols=3,
    figsize=(18,10)
)

#barplot
sns.barplot(data=df,x="Payment",y="Total",ax=ax[0,0])
ax[0,0].set_title("Payment by Method")

#boxplot
sns.boxplot(data=df,y="Total",x="Gender",color="pink",ax=ax[0,1])
ax[0,1].set_title("Expenses by Gender")

#violinplot
sns.violinplot(data=df,y="Total",x="Customer type",ax=ax[0,2],color="orange")
ax[0,2].set_title("Sales by Customer type")

#stripplot
sns.stripplot(data=df,y="Total",x="City",color="Green",jitter=0.2,ax=ax[1,0])
ax[1,0].set_title("Sales by Region")

#boxenplot
sns.boxenplot(data=df,x="Product line",y="Total",color="yellow",ax=ax[1,1])
ax[1,1].set_title("Sales by Product Category")

#swarmplot
sns.swarmplot(data=df,x="Branch",y="Total",color="red" ,ax=ax[1,2])
ax[1,2].set_title("Sales by Branch")


# Main dashboard title
fig.suptitle(
    "Mart Sales Analysis Dashboard",
    fontsize=18,
    fontweight="bold"
)

# Adjust spacing
plt.tight_layout()

# Display
plt.show()


```

4.  .set_title()


.set_title() is a Matplotlib Axes method used to add or customize the title of an individual chart

this used when creating charts using fig, ax = plt.subplots()

syntax 


ax.set_title(
    label,
    fontdict=None,
    loc=None,
    pad=None,
    *,
    y=None,
    fontsize=None,
    fontweight=None,
    color=None,
    rotation=None,
    **kwargs
)


where ,

| Parameter    | Meaning                          | Example                         |
| ------------ | -------------------------------- | ------------------------------- |
| `label`      | Title text (required)            | `"Sales Analysis"`              |
| `fontdict`   | Dictionary of font properties    | `{"fontsize":15}`               |
| `loc`        | Title position                   | `"left"`, `"center"`, `"right"` |
| `pad`        | Distance between title and chart | `pad=20`                        |
| `y`          | Vertical position of title       | `y=1.05`                        |
| `fontsize`   | Title text size                  | `fontsize=16`                   |
| `fontweight` | Title thickness                  | `"bold"`                        |
| `color`      | Title color                      | `"blue"`                        |
| `rotation`   | Rotate title text                | `rotation=45`                   |


5.  .suptitle()


.suptitle() is a Matplotlib Figure method used to add a main title for the entire figure/dashboard.

It is mainly used when you have multiple charts in one figure using plt.subplots()



Syntax

fig.suptitle(
    t,
    fontsize=None,
    fontweight=None,
    color=None,
    x=0.5,
    y=0.98,
    ha=None,
    va=None,
    **kwargs
)

where,

| Parameter    | Meaning              | Example             |
| ------------ | -------------------- | ------------------- |
| `t`          | Main title text      | `"Sales Dashboard"` |
| `fontsize`   | Title size           | `fontsize=18`       |
| `fontweight` | Title thickness      | `"bold"`            |
| `color`      | Title color          | `"blue"`            |
| `x`          | Horizontal position  | `x=0.5` (center)    |
| `y`          | Vertical position    | `y=1.02`            |
| `ha`         | Horizontal alignment | `"center"`          |
| `va`         | Vertical alignment   | `"top"`             |


6. plt.tight_layout()


plt.tight_layout() is a Matplotlib function used to automatically adjust the spacing between multiple charts in a figure


It prevents:

Titles overlapping
Axis labels getting cut off
Charts touching each other
Subplot text collision




Syntax

plt.tight_layout(
    pad=1.08,
    h_pad=None,
    w_pad=None,
    rect=None
)


where,

| Parameter | Meaning                           | Example             |
| --------- | --------------------------------- | ------------------- |
| `pad`     | Space around the complete figure  | `pad=2`             |
| `h_pad`   | Vertical spacing between charts   | `h_pad=1`           |
| `w_pad`   | Horizontal spacing between charts | `w_pad=2`           |
| `rect`    | Limit layout area                 | `rect=[0,0,1,0.95]` |
