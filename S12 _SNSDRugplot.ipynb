B.Distribution plots 

3.Rugplot 


A rug plot is a graph that represents each data value as a small line (tick) on an axis, making it easy to see the distribution, density, clusters, and spread of the data



Syntax 


sns.rugplot(
    data=None,
    *,
    x=None,
    y=None,
    height=0.025,
    expand_margins=True,
    clip_on=True,
    legend=False,
    ax=None,
    **kwargs
)


where,


| **Parameter**    | **Meaning**                                                                                          | **Example**                           |
| ---------------- | ---------------------------------------------------------------------------------------------------- | ------------------------------------- |
| `data`           | Dataset containing the values to plot.                                                               | `data=tips`                           |
| `x`              | Column name whose values are shown as tick marks on the x-axis.                                      | `x="total_bill"`                      |
| `y`              | Column name whose values are shown as tick marks on the y-axis (for a vertical rug plot).            | `y="tip"`                             |
| `height`         | Controls the length of each tick mark. Default is `0.025`.                                           | `height=0.05`                         |
| `expand_margins` | Adds extra space around the plot so rug ticks are fully visible.                                     | `expand_margins=True`                 |
| `clip_on`        | If `True`, tick marks stay inside the plot boundaries. If `False`, they can extend outside the axes. | `clip_on=False`                       |
| `legend`         | Displays a legend when multiple groups (using `hue`) are plotted.                                    | `legend=True`                         |
| `ax`             | Specifies the Matplotlib axes object on which to draw the rug plot.                                  | `ax=ax`                               |
| `**kwargs`       | Additional styling options such as `color`, `linewidth`, `alpha`, etc.                               | `color="red", linewidth=2, alpha=0.7` |


1.import all required libraries 

```python
import numpy as np
import pandas as pd 
import matplotlib.pyplot as plt
import seaborn as sns 
```

2.Import data set and specific columns 

```python
df=pd.read_excel("C:/data_analysis_using_python/supermarket_sales.xlsx")
df=df[[ 'Gender','Unit price','Quantity','Total','Payment','gross income']]
df.head()
```

3.Create a basic rugplot for one variable 

```python
plt.figure(figsize=(15,5))
sns.rugplot(data=df,x="Unit price")
```

4.Create a rugplot for two variables 

```python
plt.figure(figsize=(15,5))
sns.rugplot(data=df,x="Unit price",y="gross income",height=0.2)
```

5.Group it by a categorical variable using hue 

```python
plt.figure(figsize=(15,5))
sns.rugplot(data=df,x="Unit price",y="gross income", hue="Gender",height=0.2)
```

6.change the height of rugs in the rugplot

```python
plt.figure(figsize=(15,5))
sns.rugplot(data=df,x="Unit price",hue="Gender",height=0.5)
```

7.Combine with a KDE plot 

```python
plt.figure(figsize=(15,5))
sns.kdeplot(data=df,x="Unit price",y="gross income",hue="Gender",fill=True)
sns.rugplot(data=df,x="Unit price",y="gross income",hue="Gender",height=0.1)
```

8.Combine with a scatter plot 


```python
plt.figure(figsize=(15,5))
sns.scatterplot(data=df,x="Unit price",y="gross income",hue="Gender")
sns.rugplot(data=df,x="Unit price",y="gross income",hue="Gender")
```

9.show rugs outside of the axis 

```python
plt.figure(figsize=(15,5))
sns.scatterplot(data=df,x="Unit price",y="gross income",hue="Gender")
sns.rugplot(data=df,x="Unit price",y="gross income",hue="Gender",height=-0.03,clip_on=False)
```
