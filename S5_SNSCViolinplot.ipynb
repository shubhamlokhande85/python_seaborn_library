4.Violinplot


A Violin Plot is a categorical plot used to show the distribution of numerical data across different categories. It combines the features of a Box Plot and a Density Plot, making it easy to understand the spread and shape of the data



Syntax


sns.violinplot(
    x=None,
    y=None,
    hue=None,
    data=None,
    order=None,
    hue_order=None,
    orient=None,
    color=None,
    palette=None,
    split=False,
    inner='box',
    linewidth=None,
    scale='area'
)


where,


| Parameter   | Description                                                                                      |
| ----------- | ------------------------------------------------------------------------------------------------ |
| `x`         | Categorical variable on the x-axis.                                                              |
| `y`         | Numerical variable on the y-axis.                                                                |
| `hue`       | Creates separate violin plots based on another categorical variable.                             |
| `data`      | Dataset containing the variables.                                                                |
| `order`     | Specifies the order of categories on the axis.                                                   |
| `hue_order` | Specifies the order of hue categories.                                                           |
| `orient`    | Orientation of the plot (`'v'` for vertical, `'h'` for horizontal).                              |
| `color`     | Sets a single color for all violins.                                                             |
| `palette`   | Applies different colors to categories.                                                          |
| `split`     | Splits the violin into two halves when `hue` is used (`True` or `False`).                        |
| `inner`     | Displays information inside the violin (`'box'`, `'quartile'`, `'point'`, `'stick'`, or `None`). |
| `linewidth` | Sets the thickness of the violin border.                                                         |
| `scale`     | Controls the width of the violin (`'area'`, `'count'`, or `'width'`).                            |

1.import all required libraries

```python
import numpy as np
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns 
```

2.Import data

```python
df=pd.read_excel("C:/data_analysis_using_python/supermarket_sales.xlsx")
df.head()
```

3.Create a basic violinplot

```python
sns.violinplot(y='Total',data=df)
```

4.create a violinplot on two categorical and one numeric variable and use split

```python
#classifying payment method and also displaying density of its payment method by Gender
sns.violinplot(x="Payment",y="Total",hue="Gender",split=True,data=df)
```

5.Change the boxplot in the violinplot to horizontal lines 

```python
# To change boxplot stistical line as horizontal in violinplot use inner='quartilt' attribute
sns.violinplot(x="Payment",y="Total",hue="Gender",split=True,data=df,inner='quartile')
```

6.Draw line for each observation (rows) in violinplot

```python
#To draw each line of observation on violinplot we use inner="stick" attribute
sns.violinplot(x="Payment",y="Total",hue="Gender",split=True,data=df,inner="stick")
```

7.Change amount of smooting using band width (bw) attribute

```python
sns.violinplot(x="Payment",y="Total",hue="Gender",split=True,data=df,inner='quartile',bw_method=2)
```

8. cut the extreme values 

extreme values 

Extreme values are data points that are unusually large or unusually small compared to the rest of the data.


```python
# To extreme value from violinplot we use cut=0 attrbute
sns.violinplot(x="Payment",y="Total",hue="Gender",split=True,inner='quartile',data=df,cut=0)
```
