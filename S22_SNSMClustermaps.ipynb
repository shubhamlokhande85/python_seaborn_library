2.Cluster maps

A cluster map shows values using colors and automatically puts similar data close together.

Syntax 

sns.clustermap(
    data,
    pivot_kws=None,
    method='average',
    metric='euclidean',
    z_score=None,
    standard_scale=None,
    figsize=None,
    cbar_kws=None,
    row_cluster=True,
    col_cluster=True,
    row_linkage=None,
    col_linkage=None,
    cmap=None,
    center=None,
    annot=False,
    fmt='.2g',
    linewidths=0,
    xticklabels=True,
    yticklabels=True
)

where, 

| Parameter       | Description                                                                  | Example                                      |
|-----------------|------------------------------------------------------------------------------|----------------------------------------------|
| data            | Matrix/DataFrame containing the data to be clustered                         | data=df                                      |
| method          | Specifies the linkage method used to create clusters                        | method="average"                             |
| metric          | Specifies the distance measure used to calculate similarity                 | metric="euclidean"                           |
| cmap            | Sets the color scheme of the heatmap                                        | cmap="coolwarm"                              |
| standard_scale  | Standardizes the data; 0 standardizes rows and 1 columns                    | standard_scale=1                             |
| z_score         | Converts values into z-scores; 0 for rows and 1 for columns                 | z_score=0                                    |
| figsize         | Sets the size of the complete cluster map                                   | figsize=(10, 8)                              |
| row_cluster     | Controls whether rows should be clustered                                   | row_cluster=True                             |
| col_cluster     | Controls whether columns should be clustered                                | col_cluster=True                             |
| row_linkage     | Provides a custom linkage matrix for row clustering                         | row_linkage=linkage_matrix                   |
| col_linkage     | Provides a custom linkage matrix for column clustering                      | col_linkage=linkage_matrix                  |
| tree_kws        | Customizes the appearance of the clustering tree/dendrogram                 | tree_kws={"linewidths": 1}                   |
| linewidths      | Sets the width of lines between heatmap cells                               | linewidths=0.5                               |
| linecolor       | Sets the color of lines between heatmap cells                               | linecolor="white"                            |
| annot           | Displays the actual values inside the cells                                 | annot=True                                   |
| fmt             | Controls the number format of values displayed in cells                     | fmt=".2f"                                    |
| annot_kws       | Customizes the appearance of annotation text                                | annot_kws={"size":12}                        |
| cbar            | Shows or hides the color bar                                                | cbar=True                                    |
| cbar_kws        | Customizes the color bar                                                    | cbar_kws={"label":"Value"}                   |
| mask            | Hides selected cells from the heatmap                                       | mask=mask                                    |
| xticklabels     | Controls the labels displayed on the X-axis                                 | xticklabels=True                             |
| yticklabels     | Controls the labels displayed on the Y-axis                                 | yticklabels=True                             |
| center          | Sets the value at the center of the color map                               | center=0                                     |
| vmin            | Sets the minimum value for the color scale                                  | vmin=-1                                      |
| vmax            | Sets the maximum value for the color scale                                  | vmax=1                                       |
| fig             | Accesses the returned ClusterGrid object for further customization          | g=sns.clustermap(df)                         |

Clustering —


Clustering in data science is the process of grouping similar data items together based on their characteristics.


1. Import all required libraries 

```python
import numpy as np 
import pandas as pd
import matplotlib.pyplot as plt 
import seaborn as sns 
```

2.import data

```python
df = pd.read_excel("C:/Python_Seaborn_Library/mart_linePlot.xlsx")
df.columns=df.columns.str.lower()
df.head(10)
```

3.Convert data into matrix form of data using pivot_table()

```python
df.pivot_table(index="outlet_year",columns="outlet_size",values="sales")
```

4.create a cluster map 

```python
dfpiv=df.pivot_table(index="outlet_year",columns="outlet_size",values="sales")
sns.clustermap(dfpiv)
```

5.create a cluster map based of column cluster 

```python
sns.clustermap(dfpiv,
               row_cluster=False)
```

6.create a cluster based on row cluster 

```python
sns.clustermap(dfpiv,
               col_cluster=False)
```

7.To get values on clustermap using annot 

```python
sns.clustermap(dfpiv
               ,row_cluster=False
               ,annot=True )
```

8.Format values on clustermap using fmt 

```python
sns.clustermap(dfpiv
              ,row_cluster=False
              ,annot=True
              ,fmt='.0f')
```

9.create a clustr map using z - score 

```python
sns.clustermap(data=dfpiv
               ,row_cluster=False
               ,annot=True
               ,fmt='.0f'
               ,z_score=1)
```

10.Create a clustermap using standard scale 

```python
sns.clustermap(data=dfpiv
               ,row_cluster=False
               ,annot=True
               ,fmt= ".0f"
               ,standard_scale=0)
```
