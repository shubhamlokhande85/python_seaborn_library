6.Jointplot


jointplot() shows the relationship between two variables and also shows the distribution of each variable.

Its combination of relational plot and mariginal distribution plot


Syntax


sns.jointplot(
    data=None,
    *,
    x=None,
    y=None,
    hue=None,
    kind="scatter",
    height=6,
    ratio=5,
    space=0.2,
    dropna=False,
    xlim=None,
    ylim=None,
    color=None,
    palette=None,
    hue_order=None,
    hue_norm=None,
    marginal_ticks=False,
    joint_kws=None,
    marginal_kws=None,
    **kwargs
)


where, 


| Parameter        | Simple meaning                                                  | Example                        |
| ---------------- | --------------------------------------------------------------- | ------------------------------ |
| `data`           | Dataset/DataFrame to use                                        | `data=df`                      |
| `x`              | Variable shown on X-axis                                        | `x="hours"`                    |
| `y`              | Variable shown on Y-axis                                        | `y="marks"`                    |
| `hue`            | Divides data into groups/colors                                 | `hue="gender"`                 |
| `kind`           | Type of joint plot                                              | `kind="scatter"`               |
| `height`         | Controls overall plot size                                      | `height=7`                     |
| `ratio`          | Size of main plot compared with marginal plots                  | `ratio=5`                      |
| `space`          | Space between main and marginal plots                           | `space=0.2`                    |
| `color`          | Sets the plot color                                             | `color="blue"`                 |
| `palette`        | Sets colors for groups                                          | `palette="Set2"`               |
| `hue_order`      | Controls the order of hue categories                            | `hue_order=["Male", "Female"]` |
| `hue_norm`       | Normalizes numeric hue values                                   | `hue_norm=(0, 100)`            |
| `dropna`         | Removes missing values when `True`                              | `dropna=True`                  |
| `xlim`           | Sets X-axis limits                                              | `xlim=(0, 10)`                 |
| `ylim`           | Sets Y-axis limits                                              | `ylim=(0, 100)`                |
| `marginal_ticks` | Shows ticks on marginal plots                                   | `marginal_ticks=True`          |
| `joint_kws`      | Options for the main/joint plot                                 | `joint_kws={"s":50}`           |
| `marginal_kws`   | Options for marginal plots                                      | `marginal_kws={"bins":10}`     |
| `marker`         | Changes the marker shape                                        | `marker="o"`                   |
| `linewidth`      | Controls line thickness                                         | `linewidth=2`                  |
| `marginal`       | Controls marginal plot type in supported versions               | `marginal="hist"`              |
| `**kwargs`       | Additional arguments passed to the underlying plotting function | `s=50`                         |




1.import all required libraries 

```python
import numpy as np
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns
```

2.import dataset 

```python
df=sns.load_dataset('iris')
df.head()
```

3.Create basic jointplot 

```python
sns.jointplot(data=df,x="sepal_length",y="petal_length")
```

4. Change its kind to "KDE","His","reg","resid","Hex" plots

```python
# 1. Regression 
sns.jointplot(data=df,x="sepal_length",y="petal_length",kind="reg")
```

```python
# 2.hex
sns.jointplot(data=df,x="sepal_length",y="petal_length",kind="hex")
```

5.Grouping jointplot based on a categorical variable 

```python
sns.jointplot(data=df,x="sepal_length",y="petal_length",hue="species")
```

6.Format whole plot

```python
sns.jointplot(data=df,x="sepal_length",y="petal_length",
              height=7
              ,color="red")
```

7.format indivisul plot -

A. relational plot

B. marginal distribution plot 

```python
# A.relational plot
sns.jointplot(data=df,x="sepal_length",y="petal_length"
              ,joint_kws=dict(color="red",marker="*"))
```

```python
# B.marginal distribution plot
sns.jointplot(data=df,x="sepal_length",y="petal_length"
              ,marginal_kws=dict(color="green",kde=True))
```

8.Adjust height,ratio,space and show/hide the marginal_sticks 

```python
sns.jointplot(data=df,x="sepal_length",y="petal_length",
             height=5
             ,ratio=2
             ,space=2,marginal_ticks=True)
```

9.plot KDE and RUG plot on top of joint plot 

```python
i=sns.jointplot(data=df,x="sepal_length", y="petal_length",joint_kws=dict(color="red"))
i.plot_joint(sns.kdeplot)
i.plot_joint(sns.rugplot,height=0.2,color="red")
```

NOTE :


.plot_joint()


.plot_joint() is a method of Seaborn's JointGrid object used to draw a plot in the main/center area of a joint plot


.plot_joint() decides what graph will be drawn in the center of a JointGrid.

syntax 


g.plot_joint(
    func,
    **kwargs
)


where,

| Parameter  | Simple meaning                                     | Example           |
| ---------- | -------------------------------------------------- | ----------------- |
| `func`     | Function used to draw the main plot                | `sns.scatterplot` |
| `**kwargs` | Additional options passed to the plotting function | `color="blue"`    |
