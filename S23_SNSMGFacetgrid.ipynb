F. Multi-Plot Grid 


A Multi-Plot Grid is a collection of multiple graphs arranged together in rows and columns.

1.FacetGrid 


FacetGrid in Seaborn is used to create multiple plots in a grid, where each plot represents a different category or group of the data.

Syntax 

FacetGrid Syntax:

g = sns.FacetGrid(
    data=df,
    row=None,
    col=None,
    hue=None,
    col_wrap=None,
    height=3,
    aspect=1,
    palette=None,
    row_order=None,
    col_order=None,
    hue_order=None,
    sharex=True,
    sharey=True,
    legend_out=True,
    despine=True,
    margin_titles=False,
    xlim=None,
    ylim=None,
    subplot_kws=None,
    gridspec_kws=None
)

g.map(sns.scatterplot, "x", "y")

g.map_dataframe(
    sns.scatterplot,
    x="x",
    y="y"
)


where , 


| Parameter       | Description                                                                  | Example                                      |
|-----------------|------------------------------------------------------------------------------|----------------------------------------------|
| data            | DataFrame containing the data to be plotted                                 | data=df                                      |
| row             | Creates separate plots for each category in rows                            | row="sex"                                    |
| col             | Creates separate plots for each category in columns                         | col="species"                                |
| hue             | Uses different colors for different categories                              | hue="species"                                |
| col_wrap        | Controls how many columns are displayed before wrapping to the next row     | col_wrap=3                                   |
| height          | Sets the height of each individual plot                                     | height=4                                     |
| aspect          | Controls the width-to-height ratio of each plot                             | aspect=1.5                                   |
| palette         | Sets the colors used for different categories                               | palette="Set2"                               |
| row_order       | Specifies the order of categories shown in rows                             | row_order=["Male", "Female"]                 |
| col_order       | Specifies the order of categories shown in columns                          | col_order=["A", "B", "C"]                    |
| hue_order       | Specifies the order of categories used for different colors                 | hue_order=["Yes", "No"]                      |
| sharex          | Controls whether all plots share the same X-axis                             | sharex=True                                  |
| sharey          | Controls whether all plots share the same Y-axis                             | sharey=True                                  |
| legend_out      | Places the legend outside the grid when True                                | legend_out=True                              |
| despine         | Removes unnecessary lines/spines from the plots                             | despine=True                                 |
| margin_titles   | Places row variable names near the plot margins                             | margin_titles=True                           |
| xlim            | Sets the limits of the X-axis                                               | xlim=(0, 100)                                |
| ylim            | Sets the limits of the Y-axis                                               | ylim=(0, 50)                                 |
| subplot_kws     | Provides additional settings for each subplot                               | subplot_kws={"projection":"polar"}           |
| gridspec_kws    | Provides additional settings for the grid layout                            | gridspec_kws={"wspace":0.2}                  |
| legend          | Adds a legend to the FacetGrid                                               | g.legend()                                   |
| map             | Applies a plotting function to each subset of the data                     | g.map(sns.scatterplot, "x", "y")             |
| map_dataframe   | Applies a plotting function using a DataFrame for each subset               | g.map_dataframe(sns.scatterplot, x="x", y="y")|



1. import all required libraries 

```python
import numpy as np 
import pandas as pd
import matplotlib.pyplot as plt 
import seaborn as sns 
```

2. import data

```python
df = pd.read_excel("C:/Python_Seaborn_Library/mart_linePlot.xlsx")
df.columns=df.columns.str.lower()
df.head(10)
```

3.Get unique values from a single columns

```python
df.outlet_size.unique()
```

```python
df.outlet_location_type.unique()
```

4. Create a facetgrid based on single variable

```python
sns.FacetGrid(data=df,col="outlet_size")

```

5.providing data to facetgrid 

```python
fg= sns.FacetGrid(data=df, col="outlet_size")
fg.map_dataframe(sns.lineplot, x="outlet_year",y="sales")


```

```python
fg= sns.FacetGrid(data=df, col="outlet_size")
fg.map_dataframe(sns.scatterplot, x="outlet_year",y="sales")
```

6.styling a facetgrid 

```python
fg.map_dataframe(sns.scatterplot
                 ,x="outlet_year"
                 ,y="sales"
                 ,hue="outlet_location_type"
                 ,marker="+"
                 ,alpha=0.5
                 ,color="red"
                 ,s=100)
```

7.Create FacetGrid based on two variable 

```python
fg2=sns.FacetGrid(df, col="outlet_size",row="outlet_location_type")
```

8.Supplying data to the two variable facetgrid

```python
fg2.map_dataframe(sns.histplot, x ="sales")
```
