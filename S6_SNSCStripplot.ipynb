5.Strip plot 

A strip plot shows each data value as a separate dot, making it easy to understand the distribution of data in different categories


Syntax


sns.stripplot(x=None, y=None, hue=None, data=None,
              order=None, hue_order=None,
              jitter=True, dodge=False,
              orient=None, color=None,
              palette=None, size=5,
              edgecolor="gray", linewidth=0)



where,

| Parameter   | Description                                                                               | Example                            |
| ----------- | ----------------------------------------------------------------------------------------- | ---------------------------------- |
| `x`         | Specifies the categorical variable to display on the X-axis.                              | `x="day"`                          |
| `y`         | Specifies the numerical variable to display on the Y-axis.                                | `y="total_bill"`                   |
| `hue`       | Groups data using different colors based on another categorical variable.                 | `hue="sex"`                        |
| `data`      | The dataset containing the variables.                                                     | `data=tips`                        |
| `order`     | Sets the order of categories on the X-axis.                                               | `order=["Thur","Fri","Sat","Sun"]` |
| `hue_order` | Sets the order of hue categories.                                                         | `hue_order=["Male","Female"]`      |
| `jitter`    | Prevents overlapping dots by spreading them slightly. (`True`, `False`, or a float value) | `jitter=True`                      |
| `dodge`     | Separates hue groups so they do not overlap.                                              | `dodge=True`                       |
| `orient`    | Sets the plot orientation. `"v"` = vertical, `"h"` = horizontal.                          | `orient="h"`                       |
| `color`     | Sets a single color for all dots.                                                         | `color="red"`                      |
| `palette`   | Uses different colors for hue categories.                                                 | `palette="Set2"`                   |
| `size`      | Changes the size of the dots.                                                             | `size=8`                           |
| `edgecolor` | Sets the border color of dots.                                                            | `edgecolor="black"`                |
| `linewidth` | Sets the thickness of the dot border.                                                     | `linewidth=1`                      |

1.Import all required libraries 
 

```python
import numpy as np
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns 
```

2.import dataset 

```python
df=pd.read_excel("C:/data_analysis_using_python/supermarket_sales.xlsx")
df.head()
```

3.Create a basic strip plot 

```python
sns.stripplot(data=df,y="Total")
```

4.Expand markers in strip plot using jitter 

```python
sns.stripplot(data=df, y="Total",jitter=0.2)
```

5.Draw line around the points using linewidth 

```python
sns.stripplot(data=df,y="Total",jitter=0.2,linewidth=0.2)
```

6.include third categorical variable using hue 

```python
sns.stripplot(data=df, y="Total",x="Payment",jitter=0.2,hue="Gender",linewidth=0.2)
```

8.separate each level of hue using dodge

```python
sns.stripplot(data=df,y="Total",x="Payment",hue="Gender",dodge=True,linewidth=0.2,jitter=0.2)
```

9.Draw strip plot on the violin plot 

```python
sns.violinplot(data=df,y="Total",x="Payment",inner=None,color="grey")
sns.stripplot(data=df, y="Total",x="Payment",hue="Gender",dodge=True)
```

10.Draw stripplot on the boxplot 

```python
sns.boxplot(data=df,y="Total",x="Payment")
sns.stripplot(data=df,y="Total",x="Payment",hue="Gender",color="green")
```
