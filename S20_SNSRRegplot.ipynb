D.REGRESSION PLOTS

A regression plot in Seaborn is a graph that shows the relationship between two variables and also draws a best-fit line (regression line) to help us understand the trend



1.Regplot 

regplot() is a Seaborn function used to create a scatter plot with a regression (best-fit) line.


It helps us understand the relationship between two numerical variables and see whether they have a positive or negative trend.

Syntax 

sns.regplot(
    x=None,
    y=None,
    data=None,
    scatter=True,
    fit_reg=True,
    ci=95,
    n_boot=1000,
    order=1,
    logistic=False,
    lowess=False,
    robust=False,
    logx=False,
    x_estimator=None,
    x_bins=None,
    y_jitter=None,
    x_jitter=None,
    scatter_kws=None,
    line_kws=None
)


where , 


| Parameter     | Type               | Description                          | Example                       |
| ------------- | ------------------ | ------------------------------------ | ----------------------------- |
| `x`           | `str` / array-like | Data for the **X-axis**              | `x="hours"`                   |
| `y`           | `str` / array-like | Data for the **Y-axis**              | `y="marks"`                   |
| `data`        | DataFrame          | Dataset containing x and y           | `data=df`                     |
| `scatter`     | `bool`             | Shows/hides scatter points           | `scatter=True`                |
| `fit_reg`     | `bool`             | Shows/hides regression line          | `fit_reg=True`                |
| `ci`          | `int` / `None`     | Confidence interval percentage       | `ci=95`                       |
| `n_boot`      | `int`              | Number of bootstrap resamples for CI | `n_boot=1000`                 |
| `order`       | `int`              | Polynomial degree of regression      | `order=2`                     |
| `logistic`    | `bool`             | Uses logistic regression             | `logistic=True`               |
| `lowess`      | `bool`             | Uses LOWESS smoothing                | `lowess=True`                 |
| `robust`      | `bool`             | Uses robust regression               | `robust=True`                 |
| `logx`        | `bool`             | Fits regression using logarithm of x | `logx=True`                   |
| `x_estimator` | function           | Estimates y for each unique x        | `x_estimator=np.mean`         |
| `x_bins`      | `int` / array      | Divides x into bins                  | `x_bins=5`                    |
| `x_jitter`    | float              | Adds random noise to x points        | `x_jitter=0.2`                |
| `y_jitter`    | float              | Adds random noise to y points        | `y_jitter=0.2`                |
| `scatter_kws` | dict               | Controls scatter-point appearance    | `scatter_kws={"color":"red"}` |
| `line_kws`    | dict               | Controls regression-line appearance  | `line_kws={"color":"blue"}`   |







1. import all required libraries 

```python
import numpy as np
import pandas as pd 
import matplotlib.pyplot as plt
import seaborn as sns
```

2.Import dataset 

```python
tips=sns.load_dataset('tips')
tips.head(10)
```

3.Create a basic regplot 

```python
sns.regplot(data=tips,x="total_bill",y="tip")
```

4.1 . change the sytling the regplot 


```python
sns.regplot(data=tips , x = 'total_bill',y="tip", color ="red", marker="+")
```

4.2. Change the only line color 

```python
sns.regplot(data=tips , x = 'total_bill',y="tip", line_kws=dict(color="red") )
```

4.5. Change color only of datapoints 

```python
sns.regplot(data=tips, x= 'total_bill', y="tip",scatter_kws=dict(color="red"))
```

4.6. Hide the confidence interval (ci)

```python
sns.regplot(data=tips, x= 'total_bill', y="tip",scatter_kws=dict(color="red"),ci=None)
```

4.7.Change the number of bootstrap confidence interval

```python
sns.regplot(data=tips, x= 'total_bill', y="tip",scatter_kws=dict(color="red"),n_boot=500)
```
