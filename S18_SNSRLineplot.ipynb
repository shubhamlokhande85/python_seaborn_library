2.Lineplot 


A line plot is used to show the relationship or trend between two numerical variables. It is especially useful for displaying changes over time, such as sales over months, temperature over days, or marks across different tests.


By default, lineplot() shows the mean of the data points as a line



Syntax 


sns.lineplot(
    data=None,
    *,
    x=None,
    y=None,
    hue=None,
    size=None,
    style=None,
    units=None,
    weights=None,
    palette=None,
    hue_order=None,
    hue_norm=None,
    sizes=None,
    size_order=None,
    size_norm=None,
    dashes=True,
    markers=None,
    style_order=None,
    estimator='mean',
    errorbar=('ci', 95),
    n_boot=1000,
    seed=None,
    sort=True,
    err_style='band',
    err_kws=None,
    legend='auto',
    ax=None,
    **kwargs
)


where,

| Parameter     | Description                                                                   | Example                          |
| ------------- | ----------------------------------------------------------------------------- | -------------------------------- |
| `x`           | Column used for the **X-axis**.                                               | `x = "Month"`                    |
| `y`           | Column used for the **Y-axis**.                                               | `y = "Sales"`                    |
| `data`        | DataFrame containing the required columns.                                    | `data = df`                      |
| `hue`         | Creates different **colors/lines** based on a categorical variable.           | `hue = "Gender"`                 |
| `style`       | Gives different **line styles and markers** based on a category.              | `style = "Gender"`               |
| `size`        | Changes the **width of lines** based on a variable/category.                  | `size = "Category"`              |
| `units`       | Draws a separate line for each observation/group without aggregating.         | `units = "Student_ID"`           |
| `markers`     | Adds markers at individual data points.                                       | `markers = True`                 |
| `dashes`      | Controls whether different groups use dashed/solid line styles.               | `dashes = True`                  |
| `palette`     | Specifies the colors used for different groups.                               | `palette = "Set2"`               |
| `hue_order`   | Specifies the order of categories displayed by `hue`.                         | `hue_order = ["Male", "Female"]` |
| `style_order` | Specifies the order of categories displayed by `style`.                       | `style_order = ["A", "B"]`       |
| `estimator`   | Statistical function used when multiple observations have the same `x` value. | `estimator = "mean"`             |
| `errorbar`    | Controls the error interval shown around the estimated line.                  | `errorbar = ("ci", 95)`          |
| `sort`        | Sorts the data according to the X-axis before drawing the line.               | `sort = True`                    |
| `legend`      | Controls whether and how the legend is displayed.                             | `legend = "auto"`                |
| `ax`          | Specifies the Matplotlib Axes on which to draw the plot.                      | `ax = ax`                        |

NOTE:


### Confidence Interval (ci)


A confidence interval is a range of values that shows the uncertainty around an estimated value.


Confidence Interval is by default is 95% .


The line shows the estimated mean, while the shaded confidence band represents uncertainty around the mean.



### Bootstrap Confidence Interval

Bootstrap confidence interval is a range of values estimated by repeatedly resampling the available data, used to show the uncertainty around a statistical estimate such as the mean.

By default Bootstrap Confidence Interval is 1,000 bootstrap.


The line shows the estimated mean, while the shaded confidence band represents uncertainty around the mean. 



TRICK TO UNDERSTAND 


Line → mean/average value.


Confidence interval (shaded area) → range showing uncertainty around that mean.


Bootstrap → Seaborn estimates this range by repeatedly resampling the available data.


Wider band → more uncertainty.


Narrower band → less uncertainty.

7. Test with different confidence intervals and with different numbers of bootstraping 


```python
# confidence interval for standard deviation 
# The `ci` parameter is deprecated. Use `errorbar='sd'` for the same effect.

sns.lineplot(data=df, x="Outlet_Year",y="Sales",ci="sd")
```

```python
# confidence interval in percentage 
sns.lineplot(data=df, x="Outlet_Year",y="Sales",ci=50)
```

```python
# Using n_boot we can change bootstrap confidence interval 
sns.lineplot(data=df, x="Outlet_Year",y="Sales", n_boot=10)
```
