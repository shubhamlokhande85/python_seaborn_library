C.Relational plots 

Relational plots in Seaborn are used to show the relationship between two or more variables. They help us understand whether one variable changes when another variable changes.


| Chart            | What it does                                          | Example              |
| ---------------- | ----------------------------------------------------- | -------------------- |
| **Scatter Plot** | Shows the relationship between two variables.         | Study Hours vs Marks |
| **Line Plot**    | Shows a trend or change over time.                    | Sales over Months    |
| **Relplot**      | Creates relational plots like scatter and line plots. | Study Hours vs Marks |


1.Scatterplot 


A relational scatter plot is used to show the relationship between two numerical variables.


syntax 

sns.scatterplot(
    x=None,
    y=None,
    data=None,
    hue=None,
    style=None,
    size=None
)

where ,



| Parameter | Description                                            | Example Column |
| --------- | ------------------------------------------------------ | -------------- |
| `x`       | Column used for the **X-axis**.                        | `"Hours"`      |
| `y`       | Column used for the **Y-axis**.                        | `"Marks"`      |
| `data`    | DataFrame containing the columns.                      | `df`           |
| `hue`     | Gives different **colors** based on a category.        | `"Gender"`     |
| `style`   | Gives different **marker shapes** based on a category. | `"Gender"`     |
| `size`    | Changes the **size of dots** based on a variable.      | `"Marks"`      |



1.Import all required libraries 

```python
import numpy as np
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns 
```

2.import data

```python
df=pd.read_excel("C:/data_analysis_using_python/mart_train_sample.xlsx").sample(50)
df.head()
```

3.Create a basic relationship plot 

```python
sns.scatterplot(data=df,x="Item_MRP",y="Sales")
```

4.Grouping relational scatterplot basis on categorical variable using hue 

```python
sns.scatterplot(data=df, x ="Item_MRP",y="Sales",hue="Outlet_Size")
```

5.Grouping relational scatterplot basis on categorical variable using style

```python
sns.scatterplot(data=df,x="Item_MRP", y="Sales", style="Outlet_Size")
```

5.1.Grouping relational scatterplot basis on categorical variable using style and styling markers

```python
sns.scatterplot(data=df,x="Item_MRP", y="Sales", style="Outlet_Size",markers={"High":"^","Medium":"v","Small":"o"})
```

6.Grouping relational scatterplot basis on categorical variable using hue and style both together

```python
sns.scatterplot(data=df,x="Item_MRP",y="Sales",style="Outlet_Size",hue="Outlet_Size",markers={"High":"^","Medium":"v","Small":"o"})
```

7.Grouping relational scatterplot basis on numerical variable using hue and palette

```python
sns.scatterplot(data=df,x="Item_MRP",y="Sales",hue="Outlet_Size",palette="Reds",style="Outlet_Size",markers={"High":"^","Medium":"v","Small":"o"})
```

8.Grouping relational scatterplot basis on numerical variable using hue and size

```python
sns.scatterplot(data=df,x="Item_MRP",y="Sales",size="Tier", hue="Outlet_Size",palette="Reds",style="Outlet_Size",markers={"High":"^","Medium":"v","Small":"o"})
```

9.Change the marker size with  s argument

```python
sns.scatterplot(data=df,x="Item_MRP",y="Sales",s=150,color="red",edgecolor="black")
```
