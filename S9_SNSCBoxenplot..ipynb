A.Categorical plots

8.Boxenplot


boxenplot() (also called letter-value plot) is a Seaborn categorical plot used to visualize the distribution of numerical data across categories.

It is an advanced version of a boxplot that shows more information about the distribution, especially for large datasets


Boxplot = shows basic quartiles


Boxenplot = shows multiple quantiles (more detailed distribution)

Syntax 


sns.boxenplot(
    data=None,
    x=None,
    y=None,
    hue=None,
    order=None,
    hue_order=None,
    orient=None,
    color=None,
    palette=None,
    saturation=0.75,
    fill=True,
    dodge=True,
    width=0.8,
    gap=0,
    linewidth=None,
    linecolor=None,
    k_depth="tukey",
    outlier_prop=0.007,
    trust_alpha=True,
    native_scale=False,
    formatter=None,
    legend="auto",
    ax=None,
    **kwargs
)





where,

| Parameter      | Meaning                    | Example             |
| -------------- | -------------------------- | ------------------- |
| `data`         | Dataset/DataFrame          | `data=df`           |
| `x`            | Categorical variable       | `x="day"`           |
| `y`            | Numerical variable         | `y="total_bill"`    |
| `hue`          | Adds color grouping        | `hue="sex"`         |
| `order`        | Category order             | `["Mon","Tue"]`     |
| `hue_order`    | Hue order                  | `["Male","Female"]` |
| `orient`       | Plot direction             | `"v"` / `"h"`       |
| `color`        | Single color               | `"blue"`            |
| `palette`      | Color scheme               | `"Set2"`            |
| `saturation`   | Color intensity            | `0.75`              |
| `fill`         | Fill boxes or outline only | `True`              |
| `dodge`        | Separate hue boxes         | `True`              |
| `width`        | Width of boxes             | `0.8`               |
| `gap`          | Space between boxes        | `0.2`               |
| `linewidth`    | Border thickness           | `2`                 |
| `linecolor`    | Border color               | `"black"`           |
| `k_depth`      | Number of quantile levels  | `"tukey"`           |
| `outlier_prop` | Outlier proportion shown   | `0.007`             |
| `native_scale` | Keep numeric scale         | `True`              |
| `ax`           | Specific subplot location  | `ax=ax[0]`          |


1.Import all required libraires


```python
import numpy as np
import pandas as pd 
import matplotlib.pyplot as plt
import seaborn as sns 
```

2.import data 

```python
df=pd.read_excel("C:/data_analysis_using_python/supermarket_sales.xlsx")
df.head()
```

3.Create simple boxenplot

```python
sns.boxenplot(data=df,y="Total",x="Payment")
```

```python
# you will learn this method in next session it just demo plot 
fig, ax = plt.subplots(figsize=(8,5))

sns.boxenplot(
    data=df,
    y="Total",
    x="Payment",
    color="red",
    ax=ax
)

ax.set_title(
    "Total Spending Distribution by Payment Method",
    fontsize=14,
    fontweight="bold"
)

ax.set_xlabel("Payment Method")
ax.set_ylabel("Total Amount")

plt.show()
```
